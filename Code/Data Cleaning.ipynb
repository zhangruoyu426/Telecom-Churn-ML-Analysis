# Data Cleaning and Preprocessing

The dataset underwent several preprocessing steps to ensure data quality and model readiness. First, the TotalCharges column was converted from object to numeric data type, as preliminary inspection revealed 11 blank space entries corresponding to customers with zero-tenure who had not yet received their first bill. These missing values were imputed with zero, consistent with the business logic that new customers have no accumulated charges. Second, a consistent binary encoding scheme was established across all categorical variables. While the SeniorCitizen column was already encoded as 0/1, other binary features such as Churn used "Yes"/"No" labels; these were mapped to "Yes"/"No" (and subsequently all to 1/0 in the creation of the dataset for machine learning models) to maintain uniformity in the dataset for exploratory data analysis. Third, the customerID column was removed from the feature set, as its unique values provide no generalizable predictive information and risk inducing overfitting in tree-based models. Finally, redundant categorical values were addressed: features including OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies, and MultipleLines contained values such as "No internet service" or "No phone service," which are fully determined by the InternetService or PhoneService variable. These were collapsed to a simple "No" category to reduce multicollinearity and prevent the creation of redundant dummy variables during encoding. Collectively, these steps yielded a clean, consistently encoded dataset suitable for subsequent modeling and analysis.

## Drop customerID

```python
df = df.drop(columns=['customerID'])
```

## Convert TotalCharges to numeric and drop missing values

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.isna().sum()
```

```python
df[df['TotalCharges'].isna()]
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df.isna().sum()
```

## Convert Senior Citizen to categorical

```python
binary_map = {1: 'Yes', 0: 'No'}
df['SeniorCitizen'] = df['SeniorCitizen'].map(binary_map)
```

```python
df
```

## Fix values for no internet columns

```python
no_internet_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']
for col in no_internet_cols:
    df[col] = df[col].replace('No internet service', 'No')
```

```python
df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')
```

 ## And one for modeling (0,1)

```python
df_model = df.copy()
```

```python
binary_cols = ['SeniorCitizen',	'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 
               'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'PaperlessBilling', 'Churn']
for col in binary_cols:
    df_model[col] = df_model[col].map({'Yes': 1, 'No': 0})

multi_cat_cols = ['gender', 'InternetService', 'PaymentMethod', 'Contract']
df_model_encoded = pd.get_dummies(df_model, columns=multi_cat_cols, drop_first=True)
```

```python
# Convert all boolean columns to 0/1
bool_cols = df_model_encoded.select_dtypes(include=['bool']).columns
df_model_encoded[bool_cols] = df_model_encoded[bool_cols].astype(int)
```
